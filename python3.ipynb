```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
print("E-Commerce Data Analytics Project")
print("Project Started Successfully!")
```

```python
data = {
    'Product': ['Laptop', 'Mobile', 'Mouse',
                'Keyboard', 'Tablet', 'Headphone'],
    'Price': [50000, 15000, 500, 1000, 20000, 2000],
    'Quantity': [2, 3, 5, 4, 2, 3]
}

df = pd.DataFrame(data)
df
```

```python
df['Total_Sales'] = df['Price'] * df['Quantity']

df
```

```python
revenue = df['Total_Sales'].sum()

print("Total Revenue:", revenue)
```

```python
average = df['Total_Sales'].mean()

print("Average Sales:", average)
```

```python
df.loc[df['Total_Sales'].idxmax()]
```

```python
import pandas as pd

df['Total_Sales'] = df['Price'] * df['Quantity']

print("ECOMMERCE SALES REPORT")
print("----------------------")

print("Product Details:")
print(df)

print("\nTotal Products:", len(df))
print("Total Quantity:", df['Quantity'].sum())
print("Total Revenue:", df['Total_Sales'].sum())
print("Average Sales:", df['Total_Sales'].mean())

print("\nHighest Sales:")
print(df.loc[df['Total_Sales'].idxmax()])

print("\nLowest Sales:")
print(df.loc[df['Total_Sales'].idxmin()])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.bar(df['Product'], df['Total_Sales'])

plt.title("E-Commerce Sales Analysis")
plt.xlabel("Products")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()

print("PROJECT COMPLETED SUCCESSFULLY")
print("Total Revenue:", df['Total_Sales'].sum())
print("Best Selling Product:",
      df.loc[df['Total_Sales'].idxmax(), 'Product'])

print("Thank You!")
```
