```python
monthly = np.array([12000,15000,18000,14000,20000,25000])
print("Total:", np.sum(monthly))
print("Average:", np.mean(monthly))
```

```python
percentage = (monthly / np.sum(monthly)) * 100
print("Sales Percentage:", percentage)
```

```python
growth = np.diff(monthly)
print("Monthly Growth:", growth)
```

```python
target = 18000
result = monthly >= target
print("Target Achieved:", result)
print("Achieved Months:", np.sum(result))
```

```python
print("SALES SUMMARY")
print("Total:", np.sum(monthly))
print("Average:", np.mean(monthly))
print("Maximum:", np.max(monthly))
print("Minimum:", np.min(monthly))
print("Median:", np.median(monthly))
print("Standard Deviation:", np.std(monthly))
```
