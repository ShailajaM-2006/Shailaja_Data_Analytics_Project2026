```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd

data = {
    'Product': ['Laptop','Mobile','Mouse',
                'Keyboard','Tablet','Headphone'],
    'Price': [50000,15000,500,1000,20000,2000],
    'Quantity': [2,3,5,4,2,3]
}

df = pd.DataFrame(data)
df['Total_Sales'] = df['Price'] * df['Quantity']

print(df)
```

```python
plt.figure(figsize=(9,5))
plt.bar(df['Product'], df['Total_Sales'])
plt.title("Product Sales Analysis")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.plot(df['Product'], df['Total_Sales'],
         marker='o', linewidth=2)
plt.title("Product Sales Trend")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid()
plt.show()
```
